## Question 1

```python
print(np.quantile(quakes[(quakes.parsed_place == 'Japan') & (quakes.magType == 'mb')]['mag'], .95))
```

## Question 4
Add a column indicating if the earthquake happened in the ring of fire

```python
rof = ['Alaska', 'Antarctica', 'Bolivia', 'California', 'Canada', 'Chile', 'Costa Rica', 'Ecuador' , 'Fiji', 'Guaetemala', 'Indonesia', 'Japan', 'Kermadec Islands', 'Mexico', 'New Zealand', 'Peru', 'Philippines', 'Russia', 'Taiwan', 'Tonga', 'Washington']

quakes['is_rof'] = quakes['parsed_place'].isin(rof)
quakes.sample(n=5)
```

## Question 5
Calculate the number of earthquakes inside and outside the Ring of Fire

```python
quakes.groupby('is_rof')['is_rof'].count()
quakes['is_rof'].value_counts()
```

## Question 6
Find the tsunami count along the Ring of Fire

```python
print(quakes[quakes.is_rof]['tsunami'].sum())
```
